```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temp = [30, 32, 35, 40, 45, 50, 55, 60, 65, 70, 75, 80]

plt.plot(months, temp, marker='o', color='b', linestyle='-', linewidth=2, markersize=6)
plt.title('Average Monthly Temperature of Surat')
plt.xlabel('Months')
plt.ylabel('Temperature (°C)')
plt.grid()
plt.figure(figsize=(10, 6))
plt.show()
```
