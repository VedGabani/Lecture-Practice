```python
import matplotlib.pyplot as plt
```

```python
students = ['Alice', 'Bob', 'Charlie', 'David', 'Eva' , 'Frank', 'Grace', 'Hannah', 'Ian', 'Jack' , 'Kathy', 'Liam', 'Mia', 'Noah', 'Olivia', 'Paul', 'Quinn', 'Ryan', 'Sophia', 'Tom']
scoures = [85, 90, 78, 92, 88, 95, 80, 87, 91, 89, 84, 93, 86, 94, 82, 81, 83, 79, 77, 76]

plt.scatter(students, scoures, color='g', marker='o')
plt.title('Student Scores')
plt.xlabel('Students')
plt.ylabel('Scores')
plt.xticks(rotation=90)
plt.grid()
plt.figure(figsize=(12, 6))
plt.show()
```
