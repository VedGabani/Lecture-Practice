```python
import matplotlib.pyplot as plt
```

```python
fruits = ['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry' , 'Fig', 'Grape', 'Iceberg', 'Jackfruit']
sold = [75, 100, 125, 150, 175, 200, 225, 250, 275]

plt.pie(sold, labels=fruits, autopct='%1.1f%%', startangle=90)
plt.title('Fruit Sales Distribution')
plt.figure(figsize=(10, 6))
plt.show()
```
