```python
import matplotlib.pyplot as plt
```

```python
product = ['Product A', 'Product B', 'Product C', 'Product D' , 'Product E']
sales = [150, 200, 300, 250, 400]

plt.bar(product, sales)
plt.title('Sales of Products')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.grid()
plt.figure(figsize=(10, 6))
plt.show()
```
