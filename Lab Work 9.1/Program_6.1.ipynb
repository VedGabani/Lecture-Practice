```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
returns = [500, 700, 800, 900, 1000, 1200]
profit = [1000, 1500, 2000, 2500, 3000, 3500]

plt.stackplot(months, returns, profit, labels=['Returns', 'Profit'], colors=["Red" , "Black"])
plt.title('Monthly Profit')
plt.xlabel('Months')
plt.ylabel('Profit ($)')
plt.figure(figsize=(10, 6))
plt.show()
```
