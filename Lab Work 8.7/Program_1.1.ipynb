```python
import pandas as pd

print("Pivoting Data Frame")

sales = pd.DataFrame({ "Month" : [ "Jan" , "Jan" , "Feb" , "Feb" , "Marc" , "Marc" ] ,
                       "Product" : [ "Laptop" , "Mobile" , "Laptop" , "Mobile" , "Laptop" , "Mobile" , ] ,
                       "Sales" : [ 500000 , 300000 , 500000 , 300000 , 500000 , 300000 ]
                    })

pivot_Sales = Sales.pivot(index="Month" , columns="Product" , values="Sales")
month_order = ["Jan" , "Feb" , "Marc"]
pivot_Sales = pivot.reindex(month_order)
print(pivot_Sales)
```
